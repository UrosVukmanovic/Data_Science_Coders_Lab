In the file **marks.csv** you will find the data of several students: Name, Surname, Honors, Birth Year, and a list of grades.

Create **one** file `data.pdf`, where each of the students will be presented on a separate page:

```
[Student/Honors student - choose one, based on the "Honors" column] Name - header.
Year of birth - normal text

Marks: - smaller header
Language Arts - 6
Second Language English - 5
...
```

````python
# import libraries
from reportlab.pdfgen import canvas
from reportlab.lib.pagesizes import A4
````

````python
c = canvas.Canvas("data.pdf", pagesize=A4)
w, h = A4

with open ('marks.csv', 'r') as f:
    f.readline()

    for i, line in enumerate(f):
        if i > 0:
            c.showPage()

            row = line.strip().split(',')

            y = h - 50

            c.setFont('Helvetica-Bold', 20)

            if row[2] =='Y':
                title = "Honors Student"
            else:
                title = "Student"

            c.drawString(50, y, f"{title}: {row[0]} {row[1]}")

            y -= 40
            c.setFont('Helvetica-Bold', 14)
            c.drawString(50, y, f"Year of Birth: {row[3]}")

            y -= 40
            c.setFont('Helvetica-Bold', 16)
            c.drawString(50, y, "Marks")

            y -= 30
            c.setFont('Helvetica-Bold', 12)

            c.drawString(70, y, f"Languange Arts - {row[4]}")
            y -= 20
            c.drawString(70, y, f"Second Languange English - {row[5]}")
            y -= 20
            c.drawString(70, y, f"Second Languange German - {row[6]}")
            y -= 20
            c.drawString(70, y, f"Maths - {row[7]}")
            y -= 20
            c.drawString(70, y, f"Chemistry - {row[8]}")
            y -= 20
            c.drawString(70, y, f"Physics - {row[9]}")
            y -= 20
            c.drawString(70, y, f"Biology - {row[10]}")
            y -= 20
            c.drawString(70, y, f"Hystory - {row[11]}")
            y -= 20
            c.drawString(70, y, f"Geography - {row[12]}")
            y -= 20
c.save()
print("data.pdf created!")
````
