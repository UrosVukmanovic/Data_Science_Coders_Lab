Based on the previous task, write a function that will take the name of the file to generate, the buyer's data and the list of items on the invoice - and then generate and save the invoice in a file with the given name.

In the file `invoices data.json` you are going to find data for three invoices – read them in a loop and pass to the function to generate PDF format invoice files. Files should be named `invoice-1.pdf`, `invoice-2.pdf` etc.

```python
from reportlab.platypus import SimpleDocTemplate, Table, TableStyle
from reportlab.lib import colors
import json

# Read the JSON file with all invoices
# The file contains a dictionary with a key called 'invoices'
with open('../03_Tables/invoices data.json', 'r') as f:
    data = json.load(f)

br_racuna = 1

# Napomena:
# content je lista u koju dodajemo sve elemente koji treba da se pojave u PDF-u.
# U ovom primeru u nju dodajemo tabelu sa stavkama fakture.
for d in data['invoices']:
    doc = SimpleDocTemplate('invoice-' + str(br_racuna) + '.pdf')
    content = []
    data = []

    data.append(['No', 'Goods/Services', 'Quantity', 'Unit price', 'Total price'])

    datum = d['date']
    kupac_ime = d['buyer']['name']
    kupac_adresa = d['buyer']['address']
    kupac_nip = d['buyer']['nip']
    kupac_regon = d['buyer']['regon']

    total_cena = 0
    total_kolicina = 0
    total_ukupno = 0
    br_proizvoda = 1

    # Napomena:
    # for petlja prolazi kroz sve proizvode na fakturi.
    # Za svaki proizvod uzimamo naziv, cenu i količinu i dodajemo ga u tabelu.
    for i in d['items']:
        item_ime = i['name']
        item_cena = i['price']
        item_kolicina = i['amount']

        data.append([br_proizvoda, item_ime, item_kolicina, item_cena, int(item_kolicina) * int(item_cena)])

        br_proizvoda += 1
        total_cena += int(item_cena)
        total_kolicina += int(item_kolicina)
        total_ukupno += int(item_cena) * int(item_kolicina)

    # Napomena:
    # Posle for petlje znamo da su sve stavke obrađene.
    # Zato dodajemo završni red sa ukupnim iznosima.
    data.append(['Total', '', total_kolicina, total_cena, total_ukupno])

    t = Table(data)
    t.setStyle((TableStyle([
        ('ALIGN', (0, 0), (-1, -1), 'RIGHT'),
        ('GRID', (0, 0), (-1, -1), 0.25, colors.black),
        ('FONT', (0, 0), (-1, -1), 'Helvetica', 11)
    ])))

    # Ovde ubacujemo tabelu u listu content, pa je kasnije PDF kreira.
    content.append(t)
    doc.build(content)

    br_racuna += 1
```
