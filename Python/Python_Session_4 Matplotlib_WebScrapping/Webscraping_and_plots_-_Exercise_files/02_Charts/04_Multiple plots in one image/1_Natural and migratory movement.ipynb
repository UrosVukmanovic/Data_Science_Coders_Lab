In the file **Population in Poland - projections.xlsx** you will find a sheet **natural and migratory movement**.

Within a single figure, create two plots **closely together (next to each o other horizontally)**:
- births and deaths (two lines, based on the columns **Natural increase - Births** and **Natural decrease - Deaths** in the sheet)
- emigration and immigration (two lines, based on the columns **Foreign permanent migration - Immigration** and **Foreign permanent migration - Emigration** in the sheet)

Add legends to each graph explaining the meaning of each line.

```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_excel('Population in Poland - projections.xlsx', sheet_name = 'natural and migratory movement')
df
```

```python
df_births = df[['Natural increase - Births']]
df_deaths = df[['Natural decrease - Deaths']]
df_immigration = df[['Foreign permanent migration - Immigration']]
df_emigration = df[['Foreign permanent migration - Emigration']]
df_year = df[['Year']]
df_births
```

```python
# Create a figure with 2 subplots side-by-side

fig, (ax1, ax2) = plt.subplots(1,2, figsize=(14,5))

#------------------------
# 1) Births & Deaths
#------------------------
ax1.plot(df_year, df_births, '-.b', label='births')
ax1.plot(df_year, df_deaths, '-.r', label='deaths')

ax1.set_title("Births and Deaths")
ax1.set_xlabel("Year")
ax1.set_ylabel("Number of People")
ax1.legend()

#------------------------
# 2) Immigration & Emigration
#------------------------
ax2.plot(df_year, df_immigration, '-.b', label='immigration')
ax2.plot(df_year, df_emigration, '-.r', label='emigration')

ax2.set_title("Immigration and Emigration")
ax2.set_xlabel("Year")
ax2.set_ylabel("Number of People")
ax2.legend()

plt.tight_layout()
plt.show()

```

```python
plt.figure(figsize=(15,10))
plt.subplot2grid((2,2), (0,0))

plt.plot(df_year, df_births, '-.b', label='births')
plt.plot(df_year, df_deaths, '-.r', label='deaths')
plt.title("Natural increase - Births and Deaths")

plt.legend()

plt.subplot2grid((2,2),(0,1))

plt.plot(df_year, df_immigration, '-.b', label='immigration')
plt.plot(df_year, df_emigration, '-.r', label='emigration')
plt.title("Migraton")

plt.legend()

plt.tight_layout()
plt.show
```
