Read from the file **Population in Poland - projections.xlsx**, from the sheet **natural and migratory movement** the values of the column 'Foreign permanent migration - Balance'. Display it on a bar chart.

Under each bar add a label with the year.

```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_excel('Population in Poland - projections.xlsx', sheet_name='natural and migratory movement')
df.head()
```

```python
df.info()
```

```python
df_year = df[['Year']]
df_year.head()
```

```python
df_immigration = df[['Foreign permanent migration - Immigration']]
df_immigration.head()
```

```python
years = df['Year']
print(type(years))
immigration = df['Foreign permanent migration - Immigration']
plt.figure(figsize = (15,10))
plt.bar(years, immigration, color = 'orange')

plt.xticks(years, years.astype(str), rotation=45)
plt.title('Foreign permanent migration - Immigration')
plt.ylabel('Number of people')
plt.xlabel('Year')

plt.grid(axis='y', linestyle='--', alpha=0.5)
plt.tight_layout
```
