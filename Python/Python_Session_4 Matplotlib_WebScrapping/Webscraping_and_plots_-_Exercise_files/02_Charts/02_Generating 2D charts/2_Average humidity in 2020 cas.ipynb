Based on the data from file `Meteo data.csv`, create a plot showing the average daily relative humidity in Warsaw in 2020.

The plot should meet the following requirements:
1. The x-axis should be labeled with months.
2. Both x-axis and y-axis should have appropriate descriptions.
3. The plot should include a title.
4. The plot should include a legend.
5. The line of the plot should be red, continuous.

```python
import pandas as pd
df = pd.read_csv('../01_Data for plots/Meteo data.csv', sep=';', decimal='.')
df.info()
```

```python
# A list containing the starting days for the months
months_change = [0, 31, 60, 91, 121, 152, 182, 213, 244, 274, 305, 335]

# A list containing the names of the months
months_names = ['January', 'February', 'March', 'April', 'May', 'June',
                'July', 'August', 'September', 'October', 'November', 'December']
# Dataset containing 12 values: average for monthly measurements
df_daily_avg = df[['Average daily relative humidity']]
 
df_daily_avg.columns = ['Average daily relative humidity']

df_daily_avg
```

```python
df_daily_avg.index
```

```python
df['Average daily relative humidity']
```

```python
import matplotlib.pyplot as plt
plt.figure(figsize=(15,10))
plt.plot(df_daily_avg.index, df_daily_avg['Average daily relative humidity'], '-.r', label='Average daily relative humidity', drawstyle='steps-post')
plt.xticks(ticks=months_change, labels=months_names, rotation=45)
plt.ylabel('Relative Humidity (%)')
plt.xlabel('Months')
plt.title("Relative Humidity in Poland in 2020")
plt.legend()
plt.grid()
plt.savefig('chart_A.jpg')
plt.show()
```
