````python
from selenium import webdriver
from time import sleep

# Selenium sam pronalazi, preuzima i pokrece odgovarajuci ChromeDriver
browser = webdriver.Chrome()
browser.implicitly_wait(10) # rešava problem “sporog učitavanja stranice”
````

````python
url = 'https://yougov.com/en-gb/ratings/charities-organisations'
browser.get(url)
````

````python
sleep(1)
cookies_btn = browser.find_element('id', 'onetrust-accept-btn-handler')
cookies_btn.click()
````

````python
orgs_container = browser.find_element('css selector', '.rankings-entities-list')
orgs = orgs_container.find_elements("tag name", 'li')
orgs
````

````python
# complete here
def get_organization_data(org):
    results =[]

    for o in orgs:
        name = o.find_elements('tag name', 'span')[2].text
        link = o.find_element('tag name', 'a').get_attribute('href')

        record = {
            'name':name,
            'link':link
        }

        results.append(record)

    return results
results = get_organization_data(orgs)
results
````

````python
browser.quit()
````
