```python
import sys, subprocess
import importlib
def ensure(pip_name, import_name=None):
    import importlib
    import_name = import_name or pip_name
    try:
        importlib.import_module(import_name)
    except ModuleNotFoundError:
        subprocess.check_call([sys.executable, '-m', 'pip', 'install', pip_name])
ensure('selenium')
ensure('webdriver-manager', 'webdriver_manager')
print('Dependencies ensured (selenium, webdriver-manager). Restart kernel if needed.')
```

```python
import csv
from selenium import webdriver
browser = webdriver.Chrome()
browser.get(" http://books.toscrape.com/")
browser.implicitly_wait(10)
```

```python
container = browser.find_element('tag name', 'section')
```

```python
books = (
    container
    .find_element('tag name', 'ol')
    .find_elements('tag name', 'li')
)
books
```

```python
def parsiraj(books):
    rows = []
    for b in books:
        title = b.find_element('css selector', 'h3 a').get_attribute('title')
        price = b.find_element('css selector', '.price_color').text
        link = b.find_element('css selector', 'h3 a').get_attribute('href')
        rows.append({'title': title, 'price': price, 'link': link})
    return rows

books_info = parsiraj(books)
```

```python
browser.quit()
```

```python
with open('books.csv', 'w', newline='', encoding='utf-8') as file:
    writer = csv.DictWriter(file, fieldnames=books_info[0].keys())

    writer.writeheader()
    writer.writerows(books_info)

with open('books.csv', 'r') as file:
    for b in file:
        print (b)
```
