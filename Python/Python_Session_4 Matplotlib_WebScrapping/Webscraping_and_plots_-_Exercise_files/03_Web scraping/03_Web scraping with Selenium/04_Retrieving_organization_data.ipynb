````python
# import required libraries here
from selenium import webdriver
from time import sleep
````

````python
browser = webdriver.Chrome()
browser.implicitly_wait(10)
````

````python
url = ('https://yougov.com/en-gb/topics/not-for-profit/Macmillan_Cancer_Support?content=all')
browser.get(url)
````

````python
sleep(1)

button_cookie = browser.find_element('id','onetrust-accept-btn-handler')
button_cookie.click()

button_banner = browser.find_element('class name', 'vwo-banner__close-trigger')
button_banner.click()
````

````python
org_details = browser.find_element('css selector', '.entity-score')
org = org_details.find_elements('tag name', 'span')
values = org_details.find_elements('class name', 'value')

value_lst=[v.text for v in values]
print(value_lst)


lst=[o.text for o in org]
lst

````

````python
# complete here
def get_organization_detailed_data(url):
    keys_lst = [o.text for o in org]
    values_lst = [v.text for v in values]
    result = dict(zip(keys_lst, value_lst))
    return result
    
````

````python
# launch the browser here but do not open the website yet; remember to add `implicitly_wait`!
#browser.quit()
org_info = get_organization_detailed_data(url)
org_info
````
