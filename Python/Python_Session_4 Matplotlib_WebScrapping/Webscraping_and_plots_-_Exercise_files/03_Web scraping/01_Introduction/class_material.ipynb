```python
container = soup.select_one("[name='weekly_15']")
tracks = container.select('div.list-group-item')
for t in tracks:
    artist = t.select_one('b').text
    track_name = t.select_one('p').text

    msg = f'Artist: {artist} | track: {track_name}'
    print (msg)
```
