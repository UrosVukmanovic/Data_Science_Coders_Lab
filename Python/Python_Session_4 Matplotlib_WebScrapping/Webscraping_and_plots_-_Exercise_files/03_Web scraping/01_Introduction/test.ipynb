```python
import requests
from bs4 import BeautifulSoup
r = requests.get("https://www.music-to-scrape.org/")
soup = BeautifulSoup(r.text, 'html.parser')
```

```python
result = soup.find('div')
result
```

```python
result1 = soup.find_all('div')
result1
```

```python
result2 = soup.select_one('#cookie-banner')
result2
```

```python
result3 = soup.find_all(id='cookie-banner')
result3
```

```python
result = soup.find_all('span')
for r in result:
    print(r)
```

```python
result = soup.find('p')
print(result.prettify())
```

```python
p_tag = soup.find("p")
a_tag = p_tag.find("a")

print(a_tag["href"])
```

```python
container = soup.select_one("[name='weekly_15']")
tracks = container.select('div.list-group-item')
for t in tracks:
    artist = t.select_one('b').text
    track_name = t.select_one('p').text

    msg = f'Artist: {artist} | track: {track_name}'
    print (msg)
```
