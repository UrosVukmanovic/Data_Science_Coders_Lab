1. name and surname of the customer's contact person (edit the query that reads `customerNumber`)

````python
from psycopg2.extras import RealDictCursor

cursor = connection.cursor(cursor_factory=RealDictCursor) # govorimo cursoru da ce rezultat query biti dictionary po defaultu zato mozemo posle u print da prikazujemo rezultate preko kljuca
cursor.execute(f"SELECT 	c.contactfirstname,	c.contactlastname FROM customers c WHERE customerNumber = {customerNumber}")

task_one_row = cursor.fetchone()
contact_person = task_one_row
print(f"contact first name: {contact_person['contactfirstname']}\ncontact last name: {contact_person['contactlastname']}")

cursor.close()
````

2. what is the name of the salesperson taking care of this customer (again, edit the query to learn the employee ID - and with another query read the information about this employee)

````python
cursor = connection.cursor()
cursor.execute(f"SELECT c.salesrepemployeenumber FROM customers c WHERE c.customerNumber = {customerNumber}")

row_one = cursor.fetchone()
print(row_one)
salesrepemployenumber = row_one[0]
print(salesrepemployenumber)


````

````python
cursor.execute(f"SELECT e.firstname, e.lastname  FROM employees e WHERE employeenumber = {salesrepemployenumber}  ")

task_2_row = cursor.fetchone()
employee_name = {
    "employee_first_name": task_2_row[0],
    "employee_last_name": task_2_row[1]
}
print(f"Employee first name: {employee_name['employee_first_name']}\nEmployee last name: {employee_name['employee_last_name']}")
````

3. ids, statuses and their orders comments

````python
cursor.execute(f"SELECT o.ordernumber, o.status, o.comments FROM orders o WHERE customernumber = {customerNumber}")

task_3_row = cursor.fetchone()
order_info = {
    "order_number": task_3_row [0],
    "order_status": task_3_row [1],
    "order_comments": task_3_row [2]
}

print(f"Order number: {order_info['order_number']}\nOrder status: {order_info['order_status']}\nOrder comments: {order_info['order_comments']}")
````

4. the check numbers and the amounts this customer paid.

````python
cursor.execute(f"SELECT p.checknumber, p.amount FROM payments p WHERE customernumber = {customerNumber}")

task_4_row = cursor.fetchone()
payment_info = {
    "checknumber":task_4_row[0],
    "amount": task_4_row[1]
}

print(f"Check number of the customer: {payment_info['checknumber']}\nAmount paid by customer: {payment_info['amount']}")
````
