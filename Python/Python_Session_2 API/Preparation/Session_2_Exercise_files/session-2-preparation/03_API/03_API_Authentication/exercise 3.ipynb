Using the `requests` library retrieve the current exchange rates from [https://exchangeratesapi.io/](https://exchangeratesapi.io/) and display them as a list:

```
- <RATE 1> EUR/<CURRENCY 1>
- <RATE 2> EUR/<CURRENCY 2>
...
```

The result should be similar to:
```
- 4.5066 EUR/PLN
- 1.5319 EUR/CAD
- 0.86308 EUR/GBP
...
```


````python
import requests
import json

url = 'https://api.exchangeratesapi.io/v1/latest?access_key=2566162fc322a549c7fe3c17a850ced2'

response = requests.get(url)

data = response.json()

#print (data)
rates = data['rates']
print(rates)

def exchange_rates():
    for r, v in rates.items():
        print(f'-{v} EUR/{r}')

exchange_rates()

````
