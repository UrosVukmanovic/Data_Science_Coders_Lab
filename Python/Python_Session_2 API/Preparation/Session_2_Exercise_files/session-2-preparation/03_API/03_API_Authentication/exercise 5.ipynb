````python
cursor.execute("""
    SELECT
	c.customername,
	c.creditlimit
    FROM customers c
    ORDER BY c.creditlimit DESC
    LIMIT 10
""")

rows = cursor.fetchall()
for row in rows:
    print (row)
````

````python
url = f'https://api.exchangeratesapi.io/v1/latest?access_key=2566162fc322a549c7fe3c17a850ced2'

respond = requests.get(url)
data = respond.json()
rates = data["rates"]
print(rates)
````

````python
def creditlimit():
    for row in rows:
        customer = row[0]
        credit_limit_eur = float(row[1])

        usd = credit_limit_eur * rates["USD"]
        jpy = credit_limit_eur * rates["JPY"]
        gbp = credit_limit_eur * rates["GBP"]

        print(
            f"{customer}: "
            f"{usd:.2f} USD / "
            f"{credit_limit_eur:.2f} EUR / "
            f"{jpy:.2f} JPY / "
            f"{gbp:.2f} GBP"
        )


creditlimit()
````
