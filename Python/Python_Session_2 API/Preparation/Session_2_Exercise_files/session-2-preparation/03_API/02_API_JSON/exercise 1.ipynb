Using the `requests` library retrieve the current exchange rates served by the API at [http://api.nbp.pl/api/exchangerates/tables/a/](http://api.nbp.pl/api/exchangerates/tables/a/), and display them in a column like this:

```
AUD (Australian dollar): 2.8571 PLN
USD (United States dollar): 3.8849 PLN
```

Sometimes the API can return very complicated data. In this case the endpoint returns a table. Example data (in the example we do not show all currencies):
```
[
    {
        "effectiveDate": "2021-07-28",
        "no": "144/A/NBP/2021",
        "rates": [
            {
                "code": "THB",
                "currency": "bat (Tajlandia)",
                "mid": 0.1182
            },
            {
                "code": "XDR",
                "currency": "SDR (MFW)",
                "mid": 5.5302
            }
        ],
        "table": "A"
    }
]
``` 

To analyze the data returned from an endpoint, you can easily use the following piece of code:
```
import json
import requests

response = requests.get('http://api.nbp.pl/api/exchangerates/tables/a/')
data = response.json()

print(json.dumps(data, indent=4, sort_keys=True))
```

````python
import json
import requests

response = requests.get('http://api.nbp.pl/api/exchangerates/tables/a/')
data = response.json()

print(json.dumps(data, indent=4, sort_keys=True))
````

````python

response_ex_1 = requests.get('http://api.nbp.pl/api/exchangerates/tables/a/') # pristupi se podacima preko API endpoint
 
currency_data = response_ex_1.json() #podaci se dodaju varijabli sa json metodom (json podaci se preuzimaju sa API)
print(f"currency_data:{currency_data}")
print(type(currency_data))
print(currency_data[0])
print(type(currency_data[0]))
exchange_rates = currency_data[0]["rates"] # Gledamo sta imamo da li je dict ili lista ili kombinacija i pristupamo elementima
print(exchange_rates)
print(type(exchange_rates))
for d in exchange_rates:
    if d['code'] == 'AUD' or d['code'] == 'USD':
        print (f'{d['code']} ({d['currency']}):\t{d['mid']} PLN')
````
