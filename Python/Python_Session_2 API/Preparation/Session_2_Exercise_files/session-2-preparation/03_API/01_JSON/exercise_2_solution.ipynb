```python
import json

your_name = input("unesite svoje ime:\t")
number_of_tasks = int(input("unesite broj zadataka:\t"))

task_lista = []

i = 0
while i < number_of_tasks:
    task = input("unesite svoje zadatke:\t")
    task_lista.append(task)
    i += 1

todo = {
    'name': your_name,
    'tasks': task_lista
}

with open('todo.json', 'w') as file:
    json.dump(todo, file, indent=4, ensure_ascii=False)

with open('todo.json') as files:
    data = json.load(files)
    print(data)
```
