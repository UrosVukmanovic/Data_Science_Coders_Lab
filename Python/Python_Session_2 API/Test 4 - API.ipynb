1.  Napisati funkciju u Python-u koja proverava trenutno vreme i temperaturu u izabranom gradu (unetog sa tastature) preuzimanjem informacija sa https://openweathermap.org. 
    
    - Dokumentacija: https://openweathermap.org/current#name
    - Za pretvaranje temperature iz Kelvina (K) u Celzijuse (°C) koristi se formula (°C = K - 273.15); na primer, temperatura (291.96 K) odgovara temperaturi (18.81 °C).
    
    (2p) 

```python
# Uvozimo biblioteke za rad

import requests 
import json

city_name = input(str("Enter city name:\t")) # korisnik unosi zeljeni grad
api_key = "8e6e534cbed6c3b7db7eb79eb25ae299" # api_key preuzet sa openweather

# URL sa kog preuzimamo podatke, predajemo parametre city i api_key kao pristupni kljuc. funkcija lower ako su podaci o gradovima malim slovima 
url = f'https://api.openweathermap.org/data/2.5/weather?q={city_name.lower()}&appid={api_key}' 

def city_weather(url, city):
    respond = requests.get(url) # preuzimanje podataka sa url u json formatu
    data = respond.json() 
    weather = data["weather"][0]["main"] # raspakivanje datoteka koji su bili u json formatu (sada su dikt u listama i obratno)
    curr_temp = data["main"]["temp"]
    curr_temp_celsius = curr_temp - 273.15 # transformacija iz Kelvina u Celzijus
    print(f'Weather in {city}:\t{weather}\nCurrent temperature in Celsius is:\t{curr_temp_celsius:.2f}') # ispis sa 2 decimale

city_weather(url, city_name) # poziv funkcije

"""
    funkcija prikazuje trenutno vreme za uneti grad. Kako bi izbegli pucanje funkcije treba dodatno resiti neke edge cases. 
    Da li su neki gradovi dati sa crticom pa koristit regex; da li su gradovi uneti pod skracenicom (NYC)
    treba takodje dodati potvrdu za API respond i obezbediti da ako se dogodi problem funkcija uhvati gresku i prikaze odgovarajucu poruku (400 itd)
"""
```
