```python
# Kolona sa godinama
ages = df['Age']

# Granice intervala
bins = [18]
while bins[-1] < 100:
    bins.append(bins[-1] + 5)

# Histogram
plt.figure(figsize=(12, 6))

plt.hist(
    ages,
    bins=bins,
    edgecolor='black'
)

# Oznake na x-osi
labels = [f"{bins[i]}-{bins[i+1]-1}" for i in range(len(bins)-1)]

plt.xticks(
    [(bins[i] + bins[i+1]) / 2 for i in range(len(bins)-1)],
    labels,
    rotation=45
)

plt.title("Distribution of Voters' Ages")
plt.xlabel("Age range")
plt.ylabel("Number of voters")

plt.tight_layout()
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("../01_Data for plots/wpb_votes_list_2019.csv")

ages = df['Age']

bins = [18, 24, 29, 34, 39, 44, 49, 54, 59,
        64, 69, 74, 79, 84, 89, 94, 99, 101]

labels = [
    "18-23", "24-28", "29-33", "34-38",
    "39-43", "44-48", "49-53", "54-58",
    "59-63", "64-68", "69-73", "74-78",
    "79-83", "84-88", "89-93", "94-98",
    "99-100"
]

groups = pd.cut(
    ages,
    bins=bins,
    labels=labels,
    right=False
)

counts = groups.value_counts().sort_index()

plt.figure(figsize=(14, 6))
plt.bar(counts.index.astype(str), counts.values)

plt.title("Distribution of Voters' Ages")
plt.xlabel("Age range")
plt.ylabel("Number of voters")

plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```
