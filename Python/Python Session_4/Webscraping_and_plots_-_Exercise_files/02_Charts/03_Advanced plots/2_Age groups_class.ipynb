From the file **Population in Poland - projections.xlsx**, sheet **5-year age groups** read the data for 2021. Display it on a cumulative bar chart.

Add a legend that explains which part of the bars is data on women and which on men.

Under each bar, add a label that tells which age group the bar applies to.

```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np

# Učitavanje podataka
df = pd.read_excel(
    "../01_Data for plots/Population in Poland - projections.xlsx",sheet_name="5-year age groups")
df
```

```python
df['Year'] == 2021
```

```python
# Filtriranje za 2021. godinu
df_2021 = df[df['Year'] == 2021]
print(df_2021.shape)

age_groups = df_2021['Age']
men = df_2021['Men']
women = df_2021['Women']

x = np.arange(len(age_groups))
print(x)
plt.figure(figsize=(16, 8))

# Kumulativni (stacked) stubičasti dijagram
plt.bar(x, men, label='Men')
plt.bar(x, women, bottom=men, label='Women')

plt.title('Population in Poland by 5-year Age Groups (2021)')
plt.xlabel('Age Group')
plt.ylabel('Population')

# Oznaka svake starosne grupe
plt.xticks(x, age_groups, rotation=45)

# Legenda
plt.legend()

plt.tight_layout()
plt.show()
```
