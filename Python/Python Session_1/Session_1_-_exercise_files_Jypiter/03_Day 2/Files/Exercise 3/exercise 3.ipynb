# Exercise 3

Write a script that moves the files: `one.txt` and `two.txt` to the `archive/` folder. The file names should remain the same, only the place where they are stored is to change.


```python
from pathlib import Path
import shutil
import time

source_folder = Path(".")
archive_folder = source_folder / "archive"

archive_folder.mkdir(exist_ok = True)

one_hour_ago = time.time() - 60*60

for text_file in source_folder.glob("*.txt"):
    if text_file.stat().st_mtime < one_hour_ago: #proveravamo da li je bilo aktivnosti u zadnjih 7 dana
        shutil.move(str(text_file), str(archive_folder / text_file.name)) # za svaki txt fajl koji je stariji od 7 dana premestamo u archive folder
print("Old .txt files moved to arhcive.")
```
