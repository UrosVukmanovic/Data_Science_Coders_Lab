```python
import pandas as pd

data = {
    "Student": ["Alice", "Bob", "Charlie", "David", "Eve", "Frank", "Grace", "Helen", "Ian", "Jack"],
    "Math": [85, 90, 78, 88, 92, 76, 81, 95, 89, 84],
    "Science": [91, 85, 80, 94, 88, 79, 87, 92, 90, 83],
    "English": [88, 92, 85, 91, 93, 78, 84, 89, 86, 80],
    "History": [90, 88, 85, 91, 95, 82, 80, 87, 88, 86],
    "Art": [92, 79, 84, 87, 90, 88, 92, 85, 83, 90]
}

df = pd.DataFrame(data)
df
```

3. Write a function that calculates and returns the correlation coefficient between the `Math` and `Science` scores. Explain the correlation score. (1.5p)

```python
def correlation_coef(df):
    corr = df['Math'].corr(df['Science'])
    return corr

correlation_coef(df)

```

4. In the given dataframe `df`, add new column `total` that calculates the total score for each student (sum of scores in all subjects) and print the name of the student with the highest total score. (2p)

```python
dflist = ['Math', 'Science', 'English', 'History', 'Art'] # lista kolona sa ocenama za sum
df['total'] = df[dflist].sum(axis=1) # pravimo kolonu total tako sto sumiramo novi df - df[dflist] axis=1 =row wise sum
df_sorted = df.sort_values(by='total', ascending=False) # sortiramo prema kololin total descending - najbolji ka najlosijem rezultatu
df_sorted
```

```python
top_student = df.loc[df['total'].idxmax(), 'Student'] # filtriramo samo top studenta i prikazujemo samo ime naznakom na kolonu 'Student'
print(top_student)
```

- Bonus task: Transform the dataframe so that each student could be in several rows, but one row for each subject. Print the data, and then return it to the original format with the following transformation. (2p)

```python
# wide -> long (subjects in one column) and back to wide
subjects = ['Math', 'Science', 'English', 'History', 'Art']
# wide -> long
df_long = df.melt(id_vars='Student', value_vars=subjects, var_name='Subject', value_name='Score')
print('Long format (first rows):')
print(df_long.head().to_string(index=False))

# pivot subjects as rows, students as columns
pivot_subjects_rows = df_long.pivot(index='Subject', columns='Student', values='Score')
print('\nPivot (Subjects as rows, Students as columns):')
print(pivot_subjects_rows)

# reconstruct wide format (students as rows, subjects as columns)
df_reconstructed = df_long.pivot(index='Student', columns='Subject', values='Score').reset_index()
df_reconstructed = df_reconstructed[['Student'] + subjects]
print('\nReconstructed wide format (first rows):')
print(df_reconstructed.head().to_string(index=False))
```
