```python
df_price_carrot = df.loc[
    (df['product_types'] == 'frozen carrot and pea mix - per 1kg') &
    (df['province'] == 'POLAND') &
    (df['date'].str.startswith ('2015'))
]
df_price_carrot
```

```python
df_price_carrot['value'].mean()
```

2. What was the average price of apple juice in 2016-2018 in the Masovia province?

```python
df_price_apple = df.loc[
    (df['product_types'] == 'apple juice, boxed - per 1l') &
    (df['date'] >= '2016-1') &
    (df['date']<= '2018-1') &
    (df['province'] == 'MASOVIA')
]
df_price_apple.head()
```

```python
df_price_apple['value'].mean()
```

3. What was the average price of tomato paste in Lower Silesia Province in 2003-2015? Compared with other products does it seem reasonable to you?

```python
df_tomato_paste = df.loc[
    (df['product_types'] == '30% tomato concentrate - per 1kg') &
    (df['date'] >= '2003-1') &
    (df['date']<= '2015-1')
]
df_tomato_paste.head()
```

```python
df_tomato_paste['value'].mean()
```

```python
df_tomato_paste2 = df.loc[
    (df['product_types'] == '30% tomato concentrate - per 1kg') &
    (df['date'] >= '2003-1') &
    (df['date']<= '2015-1') &
    (df['value'] > 0)
]
df_tomato_paste2.head()
```

```python
df_tomato_paste2['value'].mean()
```
