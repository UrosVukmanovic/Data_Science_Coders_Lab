Below is the code, which contains some information about the fuel consumption of different engines. Your task is to calculate how much fuel **on average** gasoline engines use, and how much diesel engines do. Display the results using the `print` function.

```python
fuel_consumption = {  # in liter/100km
  'Citroen': {
    'Berlingo': {
      'petrol_engine': 8.25,
      'diesel_engine': 5.75,
    },
    'C3': {
      'petrol_engine': 7.0,
      'diesel_engine': 5.0,
    },
  },
}
```

```python
fuel_consumption = {  # in liter/100km
  'Citroen': {
    'Berlingo': {
      'petrol_engine': 8.25,
      'diesel_engine': 5.75,
    },
    'C3': {
      'petrol_engine': 7.0,
      'diesel_engine': 5.0,
    },
  },
}

petrol_total = 0
diesel_total = 0
count = 0

for brand in fuel_consumption.values():
    for model in brand.values():
        petrol_total += model['petrol_engine']
        diesel_total += model['diesel_engine']
        count += 1

petrol_avg = petrol_total / count
diesel_avg = diesel_total / count

print("Average petrol consumption:", petrol_avg)
print("Average diesel consumption:", diesel_avg)
```
