18. Quadratic equations
Write a program, that will help high school students count the roots of quadratic equations. The program:

is to display a message on the screen: The equation: a*x**2 + b*x + c == 0, (as a markdown cell)
takes from the user, and appropriately stores the values for variables: a, b, c,
calculates delta,
if delta > 0, calculates the x_1 and x_2 values and displays them as:
Primes of quadratic equation
x_1 = <value>
x_2 = <value>

if delta = 0, calculates the x_1 and x_2 values and displays them on the screen as:
Primes of quadratic equation
x_1 = x_2 = <value>

if the delta is negative, it will print the corresponding information on the screen.

```python
import math
a = float(input("Enter a: "))
b = float(input("Enter b: "))
c = float(input("Enter c: "))

delta = b**2 - 4*a*c

if delta > 0:
    x1 = (-b - math.sqrt(delta)) / (2*a)
    x2 = (-b + math.sqrt(delta)) / (2*a)
    
    print("Primes of quadratic equation")
    print(f"x_1 = {x1}")
    print(f"x_2 = {x2}")

elif delta == 0:
    x = -b / (2*a)
    
    print("Primes of quadratic equation")
    print(f"x_1 = x_2 = {x}")

else:
    print("The equation has no real roots (delta < 0)")
```

Using a loop.

20. Average
Write a program, that:

writes any numbers to a list stored in the numbers variable,
counts their sum and average,
writes these numbers out on the screen,
compares the average and the sum with each other and displays the result of the comparison.

```python
n = int(input("enter the size of the list:\t"))

number_list = []
i = 0

while i < n:
        x = (int(input("enter list element:\t")))
        number_list.append(x)
        i += 1
print(number_list)

total = sum(number_list)
average = total / len(number_list)

if total > average:
    print(f"{total}\tis greater than\t{average}")
else:
    print(f"{total}\tis equal to\t{average}")
```

Fizzbuzz
For numbers from 1 to 100 (inclusive), write a program that:

if the number is divisible by 3, output Fizz (for example, for the number 6, Fizz should be displayed),
if the number is divisible by 5, output Buzz (for example, for the number 10, Buzz should be displayed),
if the number is divisible by 3 and 5, output FizzBuzz (for example, for the number 15, only FizzBuzz should be displayed),
otherwise, output the number on the screen (for example for 4 the output should be 4).

```python


for i in range(101):
    if (i % 3 == 0) and (i % 5 == 0):
        print("fizzbuzz")
    elif i % 5 == 0:
        print("buzz")
    elif i % 3 == 0:
        print("fizz")
    else:
        print(i)
```
