Random forest - classification

1. We will build a random forest using data on patients diagnosed with heart disease. The model's task will be to determine whether a patient has heart disease or not.

The data we have:

    age - the patient's age
    sex - gender of the patient
    chestpain - type of chest pain.
    trestbps - resting blood pressure (in mmHg).
    chol - cholesterol level (in mg/dl)
    fbs - fasting sugar level (if greater than 120 mg/dl, value "Y", opposite "N")
    restecg - the result of an electrocardiographic test.
    minhr - minimum measured heart rate
    maxhr - maximum measured heart rate
    exang - exercise-induced angina
    oldpeak - exercise-induced ST-segment depression relative to rest ("ST" is an element on the ECG graph)
    slope - the slope of the ST segment
    ca - number of major blood vessels
    thal - determines whether thalassemia (a genetic blood disease) is present
    ahd - determines whether the patient has heart disease

2. Download the heart_data_normalized.csv file from the materials.
3. Upload the file to the runtime environment in Google Colab (or put it on Google Drive) and,      using the Pandas library, load it into a variable.
4. Import the necessary classes to build the random forest using the Scikit-learn library.
5. Split the dataset into a training set and a test set in a 70:30 ratio (set the random seed to 42).
6. Create an instance of the rf_classifier random forest model with the random_state and n_jobs parameters defined.
7. Assign a dictionary to the params variable with the hyperparameter names as keys: max_depth, min_samples_leaf and n_estimators. Assign 2 to 4 values to each key as a list.
8. Create a GridSearchCV object that you will use to select the best set of parameters from the defined params grid for the rf_classifier model. The metric we want to calculate is precision.
9. Using the best set of hyperparameters, train the model on the training data - X_train, y_train.
10. With the test data, make a prediction using the trained model.
11. Generate a report with model metrics (classification_report) for the test set.

```python
import pandas as pd

heart_data_normalized_df = pd.read_csv("Day 3/heart_data_normalized.csv")
heart_data_normalized_df.head()
```

```python
from sklearn.model_selection import train_test_split

X = heart_data_normalized_df.drop("ahd_yes",axis=1)
y = heart_data_normalized_df["ahd_yes"]

X_train, X_test, y_train, y_test = train_test_split(X, y, random_state=42, test_size=0.3)
```

```python
from sklearn.ensemble import RandomForestClassifier

rf_classifier = RandomForestClassifier(random_state=42, n_jobs=-1)
```

```python
params = {
    "max_depth": [5, 10, 15, None],
    "min_samples_leaf": [1, 2, 4],
    "n_estimators": [100, 200, 300]
}
```

```python
from sklearn.model_selection import GridSearchCV

grid_search = GridSearchCV(estimator=rf_classifier, param_grid=params, scoring='precision')
grid_search.fit(X_train, y_train)

best_rf_classifier = grid_search.best_estimator_
print(grid_search.best_params_)
```

```python
from sklearn.metrics import classification_report

y_pred = best_rf_classifier.predict(X_test)
print(classification_report(y_test, y_pred))
```
