    The stopwords will be removed from the texts. You can use any of the ways to do it you have learned.
4. To the DataFrame, add a review_cleaned column with the result of calling the function from step 3 on the column     storing the review texts. Use the .apply() method from the Pandas library.

```python
import pandas as pd

td_df = pd.read_csv("./imdb_reviews.csv")
td_df.head()
```

```python
import re
import nltk
from nltk.corpus import stopwords

nltk.download("stopwords", quiet=True)
stop_words = set(stopwords.words("english"))


def clean_review(review):
    review = review.lower()
    review = re.sub(r"[^a-z\s]", " ", review)
    words = review.split()
    words = [word for word in words if word not in stop_words]
    return " ".join(words)


td_df["review_cleaned"] = td_df["review"].apply(clean_review)
td_df[["review", "review_cleaned"]]
```

`apply()` prosleđuje svaki pojedinačni tekst iz kolone `review` funkciji `clean_review`. Funkcija zatim tekst pretvara u mala slova, pomoću regex-a uklanja sve što nisu slova ili razmaci, a `split()` deli tekst na tokene, odnosno reči. Nakon toga se uklanjaju stop reči, a preostale reči se ponovo spajaju u očišćen tekst koji se čuva u novoj koloni `review_cleaned`.
