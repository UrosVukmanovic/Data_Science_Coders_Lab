Using the loans_df frame prepared in the previous task, create visualizations using the Plotly library. Add a title for each graph, as well as X and Y axis titles.

1. A visualization showing the average property loan amount by the type of area where the property is located and the status of the loan. The status variable should be represented on the chart by color (for each type of area there should be two columns, located next to each other with different colors).


```python
import pandas as pd
import plotly.express as px

# Load and prepare the loans data
df_loans = pd.read_csv('./data/loans.csv')
df_loans = df_loans.rename(columns=str.lower)

# Fill categorical missing values with the mode
categorical_columns = [
    "gender",
    "married",
    "dependents",
    "self_employed",
    "education",
    "status",
    "area",
]

for column in categorical_columns:
    df_loans[column] = df_loans[column].fillna(df_loans[column].mode()[0])

loans_df = df_loans
```

```python
avg_loan_by_area_status = loans_df.groupby(['area', 'status'])['loan_amount'].mean().reset_index()

# Create grouped bar chart
fig = px.bar(
    avg_loan_by_area_status,
    x='area',
    y='loan_amount',
    color='status',
    barmode='group',
    title='Average Loan Amount by Area and Loan Status',
    labels={
        'area': 'Area Type',
        'loan_amount': 'Average Loan Amount',
        'status': 'Loan Status'
    },
    height=500,
    width=900
)

fig.show()
```

Task 1 - Average Loan Amount by Area & Status:
The grouped bar chart reveals how loan amounts vary across urban, semiurban, and rural areas, and whether loans were granted or denied. This helps identify geographic patterns in lending and loan sizes, showing if certain areas have higher approved loan amounts than others.

2. Visualize the relationship between the borrower's income and the loan amount.
Configure the chart so that the color of the marker is correlated with the married variable.

```python
# Task 2: Scatter plot - Relationship between borrower income and loan amount
fig2 = px.scatter(
    loans_df,
    x='applicant_income',
    y='loan_amount',
    color='married',
    title='Relationship between Borrower Income and Loan Amount',
    labels={
        'applicant_income': 'Applicant Income',
        'loan_amount': 'Loan Amount',
        'married': 'Married Status'
    },
    height=600,
    width=900,
    opacity=0.7
)

fig2.show()
```

Task 2 - Borrower Income vs Loan Amount:
The scatter plot demonstrates the relationship between applicant income and loan amount, with marital status differentiation. Key insight: there's likely a positive correlation—higher income applicants tend to receive larger loans. The color separation shows if married/unmarried status influences this relationship or loan approval patterns.

3. Visualization showing the correlation matrix of continuous numerical variables
Set the oranges color palette.

```python
# Select only numerical columns
numerical_cols = loans_df.select_dtypes(include=['number']).columns.tolist()

# Calculate correlation matrix
correlation_matrix = loans_df[numerical_cols].corr()

# Create heatmap with oranges color palette
fig3 = px.imshow(
    correlation_matrix,
    labels=dict(color="Correlation"),
    x=numerical_cols,
    y=numerical_cols,
    color_continuous_scale='Oranges',
    title='Correlation Matrix of Numerical Variables',
    height=700,
    width=800
)

fig3.update_layout(
    xaxis_title='Variables',
    yaxis_title='Variables'
)

fig3.show()
```

Task 3 - Correlation Matrix:
The heatmap identifies which numerical variables are most strongly correlated with each other. Key patterns: applicant income likely correlates with loan amount, and co-applicant income may provide additional correlation insights. This helps prioritize which variables are most influential in lending decisions and loan sizing.
