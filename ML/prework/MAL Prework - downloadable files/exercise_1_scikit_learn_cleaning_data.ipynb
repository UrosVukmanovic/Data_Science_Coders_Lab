Cleaning Data

The exercise on preparing data for the model will be based on a single dataset. The heart_data set contains diagnostic information about one of the most important organs in the human body - the heart. The heart is an organ that constantly works at a high speed, performing about 100,000 beats and pumping 7200 liters of blood - every day. Based on medical parameters and the patient's own observations, it is possible to diagnose heart diseases. Our dataset contains the following information:

    age - the patient's age
    sex - gender of the patient
    chestpain - type of chest pain.
    trestbps - resting blood pressure (in mmHg).
    chol - cholesterol level (in mg/dl)
    fbs - fasting sugar level (if greater than 120 mg/dl, value "Y", opposite "N")
    restecg - the result of an electrocardiographic test.
    minhr - minimum measured heart rate
    maxhr - maximum measured heart rate
    exang - exercise-induced angina
    oldpeak - exercise-induced ST-segment depression relative to rest ("ST" is an element on the ECG graph)
    slope - the slope of the ST segment
    ca - number of major blood vessels
    thal - determines whether thalassemia (a genetic blood disease) is present
    ahd - determines if the patient has heart disease

```python
import pandas as pd

heart_df = pd.read_csv("data/heart_data.csv")
heart_df.tail(15)
```

```python
# droping minhr column (all Null values)
heart_df = heart_df.drop(columns=["minhr"])
heart_df
```

```python
heart_df.info()
```

```python
# check for duplicates
print(f"Number of duplicate rows before removal: {heart_df.duplicated().sum()}")
heart_df = heart_df.drop_duplicates()
print(f"Number of duplicate rows after removal: {heart_df.duplicated().sum()}")
heart_df
```

Blank values

```python
heart_df.info()
```

```python
heart_df.isnull().sum()
```

```python
from sklearn.impute import SimpleImputer
imputer = SimpleImputer(strategy="median")

cat_cols = ["thal", "sex", "chestpain", "fbs", "exang", "ahd"]

heart_df_categorical = heart_df[cat_cols]

heart_df_numeric = heart_df.drop(columns=cat_cols)
heart_df_numeric = pd.DataFrame(
    imputer.fit_transform(heart_df_numeric),
    columns=imputer.get_feature_names_out()
)

heart_df = pd.concat([heart_df_numeric, heart_df_categorical], axis=1)
heart_df


```

```python
heart_df.info()
```

Verify that values in categorical variables do not have errors. If there are inaccuracies, correct them.

```python
# column thal has 2 categorical missing values - deleted
heart_df = heart_df.dropna()
heart_df.info()
```

```python
# Save the cleaned dataset to a CSV file without the index column
heart_df.to_csv("data/heart_data_cleaned.csv", index=False)
print("Cleaned dataset saved to data/heart_data_cleaned.csv")
```
