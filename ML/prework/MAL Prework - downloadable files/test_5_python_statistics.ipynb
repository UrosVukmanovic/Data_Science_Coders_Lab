1. You have a dictionary real_estate_prices, which stores the prices of properties in Manhattan, New York. The keys are the street names, and the values are the property prices in dollars.

```python
real_estate_prices = {
    'Broadway': 1200000,
    'Park Avenue': 3500000,
    'Madison Avenue': 2800000,
    'Fifth Avenue': 5000000,
    'Lexington Avenue': 2100000,
    'West End Avenue': 1800000,
    'Central Park West': 4500000,
    'Riverside Drive': 1600000,
    'Waverly Place': 1100000,
    'Bleecker Street': 900000,
    'Christopher Street': 950000,
    'Houston Street': 1300000,
    'Canal Street': 800000,
    'Bowery': 700000,
    'Delancey Street': 1000000
}
```

```python
import numpy as np
```

```python
rel_price = [ v for k,v in real_estate_prices.items()]
rel_price = np.array(rel_price)
print(type(rel_price))
```

2. From the dictionary, extract the property prices, which you store in a NumPy array.
    1. Calculate:

        1. Mean
        2. Median
        3. The minimum and maximum values
        4. Variance
        5. Standard deviation
        


```python
mean = np.mean(rel_price)
median = np.median(rel_price)
min = np.min(rel_price)
max = np.max(rel_price)
var = np.var(rel_price)
standar_dev = np.std(rel_price)

print(mean)
print(median)
print(min)
print(max)
print(var)
print(standar_dev)
```
