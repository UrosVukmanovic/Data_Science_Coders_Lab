Preparing and Cleaning Data

```python
import pandas as pd

df_loans = pd.read_csv('./data/loans.csv')
df_loans
```

```python
df_loans.columns
```

```python
df_loans = df_loans.rename(columns=str.lower)
df_loans.head(10)

```

```python
df_loans.info()
```

```python
print(df_loans["dependents"].unique())
print(df_loans["gender"].unique())
print(df_loans["education"].unique())
print(df_loans["self_employed"].unique())
```

```python
missing_values = df_loans.isnull().sum()
print(missing_values)
```

Fixing Data Types

```python
# Fill categorical missing values with the mode
categorical_columns = [
    "gender",
    "married",
    "dependents",
    "self_employed",
    "education",
    "status",
    "area",
]

for column in categorical_columns:
    df_loans[column] = df_loans[column].fillna(df_loans[column].mode()[0])

# Check missing values after cleaning
print(df_loans.isnull().sum())
```

```python
# Columns still having null values
print(df_loans[["term", "credit_history"]].isnull().sum())
```

```python
df_loans[categorical_columns] = df_loans[categorical_columns].astype("string")
df_loans.info()
```

Filtering Data

```python
loans_granted_df = df_loans.loc[
    (df_loans["status"]) == "Y"
]
print(loans_granted_df)
loans_granted_count = loans_granted_df["gender"].count()
print(f"Number of loans granted: {loans_granted_count}")
```

```python
standalone_borrowers_df = df_loans.loc[
    (df_loans["status"] == "Y") &
    (df_loans["dependents"] == str(0))
]
standalone_borrowers_df
```

Grouping Data for Insights

```python
loans_by_gender = df_loans[['gender','applicant_income','loan_amount']].groupby(["gender"]).agg({
    'gender': 'count', 
    'applicant_income': 'mean',
    'loan_amount': 'mean'
    })
loans_by_gender.info()
loans_by_gender= loans_by_gender[['gender', 'applicant_income', 'loan_amount']].astype("int")
loans_by_gender
```
