Write a function that takes three arguments: a, b and math_operation, where a and b are real numbers and math_operation is one of the four basic mathematical operations: add, sub, mul, div - addition, subtraction, multiplication and division. For mathematical operations that require a specific order of arguments, assume that the larger number of the pair (a, b) comes first. The function should return the result of the declared mathematical operation for the indicated pair of numbers. Round the result to two decimal places.

```python
def calculator(a, b, math_operation):
    
    if math_operation == '+':
        result = a + b
    elif math_operation == '-':
        result = a - b
    elif math_operation == '*':
        result = a * b
    elif math_operation == "/":
        result == a / b
    return result

a = float(input("Enter first number:\t"))
b = float(input("Enter second number:\t"))
math_operation = input("Enter simple math operations: +, -, *, /")

calculator(a,b,math_operation)
```

2. Write a function that takes as an argument a list containing first names. The function should return only those names that begin with the letter C. Test the function on the list royal_family_names = ["Elizabeth", "Phillip", "Charles", "Camilla", "William", "Catherine", "Anne", "Harry", "Meghan"].

```python
def first_name(names):
    chosen_name = []

    for n in names:
        for l in n :
            if l == 'C':
                chosen_name.append(n)
    return chosen_name

royal_family_names = ["Elizabeth", "Phillip", "Charles", "Camilla", "William", "Catherine", "Anne", "Harry", "Meghan"]

first_name(royal_family_names)

```

```python
# with function .startswith
def first_name(names):
    return [name for name in names if name.startswith("C")]

royal_family_names = ["Elizabeth", "Phillip", "Charles", "Camilla", "William", "Catherine", "Anne", "Harry", "Meghan"]

first_name(royal_family_names)
```

```python
# with list index
def first_name(names):
    return [name for name in names if name[0] == "C"]

royal_family_names = ["Elizabeth", "Phillip", "Charles", "Camilla", "William", "Catherine", "Anne", "Harry", "Meghan"]

first_name(royal_family_names)
```

3. Write a function whose purpose is to return a dictionary of employees who earn more than the amount specified in the function call. The function takes two arguments:

a dictionary containing data about the employees, where the key is the employee's name and the value is a tuple storing information about the employee's position and salary
a number that specifies the amount above which the employee will be returned as a result of the function.

```python
def employee_info(employee, earnings):

    above_earning = {}
    for name, info in employee.items():
        department, salary = info
        if salary > earnings:
            above_earning[name] = info
            print(f"{name} works in {department} and earns {salary}.")

    return above_earning

employee = {"John Doe": ("Engineering",  50000),
            "Joe Black": ("Marketing", 60000),
            "Jane Smith": ("Sales", 55000)}

earnings_mark = 52000

employee_info(employee, earnings_mark)
```

```python
# dictionary comprehension:
def employee_info(employee, earnings):
    return {
        name: info
        for name, info in employee.items()
        if info[1] > earnings
    }

employee = {"John Doe": ("Engineering",  50000),
            "Joe Black": ("Marketing", 60000),
            "Jane Smith": ("Sales", 55000)}

earnings_mark = 52000

employee_info(employee, earnings_mark)
```
