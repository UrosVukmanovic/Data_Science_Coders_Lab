1. Import:
from the datasets module, a method that will allow you to create a synthetic set for the classification problem (for a hint, see the article Introduction to Scikit-learn),
from the model_selection module, a method that will divide the dataset into training and test datasets,
from the neighbors module, a class that will allow you to create an instance of the k-nearest neighbors model,
from the model_selection module, a method that will allow you to calculate the effectiveness of the model (accuracy).
2. Create a dataset according to the following parameters:
Number of samples: 1000,
Number of variables: 10,
Number of significant variables: 6.
3. Split the dataset using the 70:30 (training:testing) ratio.
4. Create an instance of the k-nearest neighbor model, specifying the number of neighbors to be 4.
5. Train the model on the training set, and then make a prediction on the test set.
6. Based on the prediction of the model and the actual labels (results) from the test subset, calculate the value of the accuracy metric.

```python
from sklearn.datasets import load_iris, make_classification
from sklearn.model_selection import train_test_split
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import accuracy_score

# Step 1: Import the libraries required for a classification problem.
# - make_classification: creates synthetic labeled data
# - train_test_split: splits data into training and testing sets
# - KNeighborsClassifier: builds the k-nearest neighbors model
# - accuracy_score: measures model accuracy

# Step 2: Generate a synthetic classification dataset.
# We create 1000 samples, 10 features, and 6 informative features.
data_classification = make_classification(
    n_samples=1000,
    n_features=10,
    n_informative=6,
    random_state=42
)
```

```python
# Step 3: Unpack the generated dataset into features (X) and labels (y).
# In scikit-learn, X contains the input values and y contains the target classes.
X_class, y_class = data_classification

# Step 4: Split the dataset into training and testing subsets.
# test_size=0.3 means 30% of the data is used for testing and 70% for training.
# random_state=42 ensures the split is reproducible.
X_train_class, X_test_class, y_train_class, y_test_class = train_test_split(
    X_class,
    y_class,
    test_size=0.3,
    random_state=42
)
```

```python
# Step 5: Create an instance of the k-nearest neighbors classifier.
# n_neighbors=4 means the model looks at the 4 closest data points when predicting.
knn = KNeighborsClassifier(n_neighbors=4)
```

```python
# Step 6: Train the model using the training data.
# The model learns patterns from X_train_class and the corresponding labels y_train_class.
knn.fit(X_train_class, y_train_class)
```

```python
# Label probability
y_pred_proba = knn.predict_proba(X_test_class)
```

```python
# Step 7: Predict classes for the test set.
# The model uses the features in X_test_class to predict labels.
# We use predict() because we need class labels, not probabilities.
y_pred = knn.predict(X_test_class)
```

```python
# Step 8: Evaluate the model using the test set.
# knn.score() compares the predicted labels against the true labels from the test set.
# It returns the accuracy as a float between 0 and 1.
knn.score(X_test_class, y_test_class)
```

```python
# Step 9: Calculate accuracy explicitly using accuracy_score.
# This function compares the actual labels with the predicted labels.
# It is the standard metric for classification accuracy.
accuracy_score(y_test_class, y_pred)
```

# Step-by-step explanation of the workflow

This notebook demonstrates a basic machine learning classification pipeline using scikit-learn.

1. Import libraries:
   - `make_classification` creates a synthetic dataset with a target label.
   - `train_test_split` divides the data into training and testing subsets.
   - `KNeighborsClassifier` creates the k-NN model.
   - `accuracy_score` measures how well the model predicts.

2. Create the dataset:
   - The dataset has 1000 samples and 10 features.
   - 6 of those features are informative, meaning they strongly influence the target label.
   - `random_state=42` keeps the data generation reproducible.

3. Split the dataset:
   - `test_size=0.3` means 30% of the data is reserved for testing.
   - The remaining 70% is used to train the model.

4. Build the model:
   - `KNeighborsClassifier(n_neighbors=4)` creates a model that predicts based on the 4 nearest examples.

5. Train the model:
   - `knn.fit(X_train_class, y_train_class)` teaches the model the relationship between features and labels using the training data.

6. Make predictions:
   - `knn.predict(X_test_class)` generates predicted labels for the unseen test data.

7. Evaluate accuracy:
   - `accuracy_score(y_test_class, y_pred)` compares the true labels with the predictions.
   - The result tells us the proportion of correctly classified examples.

This is the standard scikit-learn workflow for supervised classification:
create data -> split data -> train model -> predict -> evaluate.
