Scikit-learn - exercise

1. Import:
from the datasets module, a method that will allow you to create a synthetic set for the classification problem (for a hint, see the article Introduction to Scikit-learn),
from the model_selection module, a method that will divide the dataset into training and test datasets,
from the neighbors module, a class that will allow you to create an instance of the k-nearest neighbors model,
from the model_selection module, a method that will allow you to calculate the effectiveness of the model (accuracy).
2. Create a dataset according to the following parameters:
Number of samples: 1000,
Number of variables: 10,
Number of significant variables: 6.
3. Split the dataset using the 70:30 (training:testing) ratio.
4. Create an instance of the k-nearest neighbor model, specifying the number of neighbors to be 4.
5. Train the model on the training set, and then make a prediction on the test set.
6. Based on the prediction of the model and the actual labels (results) from the test subset, calculate the value of the accuracy metric.
