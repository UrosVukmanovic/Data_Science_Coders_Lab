Dataset

Preparing data for the model
1. Reading the dataset

```python
import pandas as pd
cancer_df = pd.read_csv("data/cancer_data_course.csv")

cancer_df.head(10)
```

2. Removing redundant variables

```python
cancer_df = cancer_df.drop(columns=["id", "Unnamed: 32", "Unnamed: 0"])
cancer_df.head()
```

3. Removing missing values


```python
# Calculating the percentage of null values

missing_values = cancer_df.isnull().sum()/(len(cancer_df))*100
missing_values.round(2)

# Maximum missing_values_% is 1.58 - it's okay to delete observations with missing values
```

```python
# Drop rows with at least 1 missing value

cancer_df = cancer_df.dropna(axis=0).reset_index()
cancer_df.info()
```

Handling blank values

```python
# Step 1: Create an imputer that fills missing numeric values with the mean of each column.
# This is a common strategy for numerical features before training a machine learning model.
from sklearn.impute import SimpleImputer
imputer = SimpleImputer(strategy="mean")

# Step 2: Save the diagnosis column separately.
# We cannot fill missing values in a text column using the mean, so we remove it from the numeric data.
diagnosis_df = cancer_df["diagnosis"]

# Step 3: Remove the target column from the feature matrix and fill missing values.
# After this, cancer_df contains only the numeric feature columns with missing values replaced by averages.
cancer_df = imputer.fit_transform(cancer_df.drop("diagnosis", axis=1))

# Step 4: Convert the NumPy array returned by imputer back into a DataFrame.
# We restore the original column names so the data is easy to work with.
cancer_df = pd.DataFrame(cancer_df, columns=imputer.get_feature_names_out())

# Step 5: Add the diagnosis column back to the cleaned data.
# The extra index column created earlier is removed because it is not needed in the final dataset.
cancer_df = pd.concat([cancer_df, diagnosis_df], axis=1).drop("index", axis=1)

# Step 6: Remove the diagnosis column from the feature DataFrame. 
# This final version keeps only the model inputs and removes the target label from the features.
cancer_df.head()


```

[RECOMMENDED] Using the .get_dummies() method, specifying the data frame and columns to be encoded. In addition, we set the drop_first argument to True. Dealing with two unique values, we would get two columns with postfixes: diagnosis_b and diagnosis_m. If the value of the diagnosis variable is "m", then diagnosis_b takes the value 0, and diagnosis_m takes the value 1. Otherwise - the variables get opposite values. However, there is no need to have two variables, because the information from one column (diagnosis_m) is sufficient. If our variable takes the value of 1, we have a malignant tumor, and with 0 - a benign one.

```python
# Encode categorical variable using get_dummies()
cancer_df = pd.get_dummies(data=cancer_df, columns=['diagnosis'], drop_first=True)
cancer_df.head(5)
```

7. Splitting the data into a training set and a test set 

We have one last step related to preparing the data for the model, namely dividing the data into subsets. In our case, we have x observations available, which is not a large number. Therefore, we will not create the validation set and will focus on dividing the set into two: training and test. For this we will use the train_test_split method, in which we will give the arguments:

the part to be the test set relative to the original set (test_size). The remainder (1 - test_size) will be the training set.
random_state that allows for repeatable results. For example, with random_state set to 42, each call to the train_test_split method will produce exactly the same split between the training and test sets. That is, the sets will always contain the same observations.
First, however, we need to separate the independent variables and the the dependent variable into separate variables:

```python
# Exclude the encoded diagnosis column from X because it is the target (the answer to predict).
# Keep it separately as y so the model learns from the other columns without seeing the answer.
X = cancer_normalized_df.drop("diagnosis_m", axis=1)
y = cancer_normalized_df["diagnosis_m"]

```

```python
# Split data into training and test set (70% training set, 30% test set)
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X,y, test_size=0.3, random_state=42)
print(X_train.shape, X_test.shape)
```

In case we would like to divide the set into three subsets (i.e., include the validation set), we would have to perform the splitting again on the training or test set:

```python
# Split train data into valid and test set (70% training set, 30% valid set). In result we have: 70% training set, 15% valid set, 15% test set
X_test, X_valid, y_test, y_valid = train_test_split(X_test, y_test, test_size=0.5, random_state=42)
print(X_train.shape, X_valid.shape, X_test.shape)
```
