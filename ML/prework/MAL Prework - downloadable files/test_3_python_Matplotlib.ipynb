1. Using the loans_df frame prepared in the previous task, make visualizations using the Matplotlib library:
    1) A column chart showing the number of borrowers by gender.
    2) Set the color of the columns to light blue.
    3) Set the title of the chart and the titles of the X and Y axes. Optionally: add labels with individual numerical values above the columns

```python
import pandas as pd
import matplotlib.pyplot as plt

# Load and prepare the loans data
df_loans = pd.read_csv('./data/loans.csv')
df_loans = df_loans.rename(columns=str.lower)

# Fill categorical missing values with the mode
categorical_columns = [
    "gender",
    "married",
    "dependents",
    "self_employed",
    "education",
    "status",
    "area",
]

for column in categorical_columns:
    df_loans[column] = df_loans[column].fillna(df_loans[column].mode()[0])

loans_df = df_loans
```

```python
borrowers_by_gender = loans_df['gender'].value_counts().sort_index()

# Create the bar chart
plt.figure(figsize=(10, 6))
bars = plt.bar(borrowers_by_gender.index, borrowers_by_gender.values, color='lightblue', edgecolor='black')

# Add titles
plt.title('Number of Borrowers by Gender', fontsize=14, fontweight='bold')
plt.xlabel('Gender', fontsize=12)
plt.ylabel('Number of Borrowers', fontsize=12)

# Add value labels on top of columns
for bar in bars:
    height = bar.get_height()
    plt.text(bar.get_x() + bar.get_width()/2., height,
            f'{int(height)}',
            ha='center', va='bottom', fontsize=11)

plt.grid(axis='y', alpha=0.3)
plt.tight_layout()
plt.show()
```

2. Visualization showing the number of borrowers by loan status (granted / not granted).
    1. Set the color of the columns to orange.
    2. Set the title of the chart and the X and Y axes.

```python
borrowers_by_status = loans_df['status'].value_counts().sort_index()

# Create the bar chart
plt.figure(figsize=(10, 6))
bars = plt.bar(borrowers_by_status.index, borrowers_by_status.values, color='orange', edgecolor='black')

# Add titles
plt.title('Number of Borrowers by Loan Status', fontsize=14, fontweight='bold')
plt.xlabel('Loan Status (Y=Granted, N=Not Granted)', fontsize=12)
plt.ylabel('Number of Borrowers', fontsize=12)

# Add value labels on top of columns
for bar in bars:
    height = bar.get_height()
    plt.text(bar.get_x() + bar.get_width()/2., height,
            f'{int(height)}',
            ha='center', va='bottom', fontsize=11)

plt.grid(axis='y', alpha=0.3)
plt.tight_layout()
plt.show()
```

3. Visualization showing the distribution of the variable loan_amount
    1. Divide the data into 50 bins, according to which you will generate a histogram.
    2. Set the color of the columns to red.
    3. Set the title of the graph and the titles of the X and Y axes.

```python
plt.figure(figsize=(12, 6))

plt.hist(loans_df['loan_amount'], bins=50, color='red', edgecolor='black')

# Add titles
plt.title('Distribution of Loan Amount', fontsize=14, fontweight='bold')
plt.xlabel('Loan Amount', fontsize=12)
plt.ylabel('Frequency', fontsize=12)

plt.grid(axis='y', alpha=0.3)
plt.tight_layout()
plt.show()
```
