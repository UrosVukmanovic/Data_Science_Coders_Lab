Using the wine parameter dataset, which is available from the datasets module of the scikit-learn library, create a data frame for which you will calculate the indicated statistical measures. The dataset consists of 13 variables:

    alcohol - alcohol content of the wine (in percentage of volume).
    malic_acid - malic acid content in wine (in g / l).
    ash - the total amount of ash in the wine (in g / l).
    alcalinity_of_ash - degree of alcalinity of ash in wine (in mEq / l).
    magnesium - the content of magnesium in wine (in mg / l).
    total_phenols - the total amount of phenols in the wine (in mg / l).
    flavanoids - total amount of flavanoids in wine (in mg / l).
    nonflavanoid_phenols - total amount of non-flavonoid phenols in wine (in mg / l).
    proanthocyanins - total amount of proanthocyanins in wine (in mg / l).
    color_intensity - color intensity of the wine (in absorbance at 520 nm).
    hue - the hue of the wine, expressed in degrees.
    od280/od315_of_diluted wines - the ratio of wine absorbance at 280 nm to absorbance at 315 nm, a variable that determines color intensity.
    proline - the amount of proline in the wine (in mg / l).

1. For each variable in the wine_df frame, calculate the mean values and the median.
2. For the alcohol and malic_acid variables, determine the minimum value, maximum value and standard deviation.
3. For the variables total_phenoles and proline calculate the quantiles - first and third.
4. Calculate the correlation matrix, and then plot the results on a visualization made with the seaborn or plotly
5. library. Take care that the graph is appropriately sized so that the labels with correlation values are readable.

```python
from sklearn.datasets import load_wine
import pandas as pd

wine_data = load_wine()
wine_df = pd.DataFrame(data=wine_data.data, columns=wine_data.feature_names)
```

```python
wine_df.head()
```

```python
df_wine_mean = wine_df.mean()
df_wine_mean
```

```python
max_values = wine_df[["alcohol", "malic_acid"]].max()
min_values = wine_df[["alcohol", "malic_acid"]].min()
std_dev = wine_df[["alcohol", "malic_acid"]].std()
print(f"Max values:\t{max_values}\n")
print(f"min:\t{min_values}\n")
print(f"Max values:\t{std_dev}")
```

```python
quartiles = wine_df[["total_phenols", "proline"]].quantile([0.25, 0.75])
quartiles
```

```python
wine_df_correlation_matrix = wine_df.corr(method="pearson")
wine_df_correlation_matrix
```

```python
import plotly.express as px
fig = px.imshow(
    wine_df_correlation_matrix,
    x = wine_df_correlation_matrix.columns,
    y = wine_df_correlation_matrix.index,
    title= "wine_df correlation matrix",
    width=700,
    height=700
)
fig.show()
```
