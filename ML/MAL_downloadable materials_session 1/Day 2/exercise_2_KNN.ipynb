```python
# Import libraries and load data into DataFrame
import pandas as pd
import seaborn as sns
from sklearn.model_selection import train_test_split
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import confusion_matrix, accuracy_score, precision_score, recall_score, f1_score

penguins_df = pd.read_csv("data/penguins_size.csv")

# Check 10 first rows of DataFrame
penguins_df.head(10)
```

```python
penguins_df.info()
```

```python
penguins_normalized_df = pd.read_csv("data/penguins_df_normalized.csv")
penguins_normalized_df.head()
```

```python
penguins_df = penguins_df.dropna()
penguins_df.info()
```

```python
penguins_df[["sex", "island", "species"]].apply(
    lambda column: column.unique()
)
```

```python
penguins_df = penguins_df[penguins_df["sex"].isin(["MALE", "FEMALE"])]
```

```python
penguins_df[["sex"]]. apply(
    lambda columns: columns.value_counts()
)
```

```python
penguins_numerical = penguins_df.columns.drop(["sex", "species", "island"])
print(penguins_numerical)

from sklearn.preprocessing import MinMaxScaler

scaler = MinMaxScaler()
penguins_df[penguins_numerical] = scaler.fit_transform(penguins_df[penguins_numerical])
print(penguins_df)
```

```python
# categorical values

penguins_categorical = penguins_df.columns.drop(["culmen_length_mm",  "culmen_depth_mm",  "flipper_length_mm", "body_mass_g", "species"])
print(penguins_categorical)

penguins_df[penguins_categorical] = penguins_df[penguins_categorical].replace({
    "Torgersen": 0,
    "Biscoe": 1,
    "Dream": 2,
    "MALE": 1,
    "FEMALE": 2
})

penguins_df.head()
```

```python
# Split data into training and test dataset
X = penguins_df.drop("species", axis=1)
y = penguins_df["species"]

X_train, X_test, y_train, y_test = train_test_split(X,
                                                    y,
                                                    test_size=0.3,
                                                    random_state=42,
                                                    stratify=y)

# Initialize k nearest neighbors and fit training data
knn = KNeighborsClassifier(n_neighbors=5)
knn.fit(X_train, y_train)

# Predict labels based on test data
y_pred = knn.predict(X_test)
print(f"KNN score: {knn.score(X_test, y_test)}")
```

The `species` column is the target (`y`); all other columns are the input features (`X`).
