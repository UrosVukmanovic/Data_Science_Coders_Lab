Real estate prices in King County

```python
import pandas as pd

housing_df = pd.read_csv("data/kc_house_data.csv")

housing_df.head(10)
```

```python
housing_df.info()
```

```python
# Droping irrelevant columns for model

housing_df = housing_df.drop(["id", "date"], axis=1)
housing_df.head()
```

```python
# Split train test set

from sklearn.model_selection import train_test_split

X = housing_df.drop("price", axis=1)
y = housing_df["price"]

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)
```

```python
from sklearn.linear_model import LinearRegression

# Initialize object linear regression model

lin_reg_model = LinearRegression()
```

```python
# Build and train model with training set of data

lin_reg_model.fit(X_train, y_train)
```

```python
# The order of the coefficients coincides with the order of the variables in the data set
print(f"Coefficient values: {lin_reg_model.coef_}")

# Intercept coefficient
print(f"Intercept coefficient value: {lin_reg_model.intercept_}")
```

```python
# Predict y values based on features in test set

y_pred = lin_reg_model.predict(X_test)
```

```python
# Print R2 score value with .score() method
print(f"R2 score: {lin_reg_model.score(X_test, y_test)}") # Or X_train, y_train for training set score

# Alternatively -R2 score using r2_score from sklearn.metrics
from sklearn.metrics import r2_score

print(r2_score(y_test, y_pred))
```
