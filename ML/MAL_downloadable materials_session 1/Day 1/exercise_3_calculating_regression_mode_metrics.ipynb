```python
import pandas as pd
kc_house_preprocessed_df = pd.read_csv("data/kc_house_data_preprocessed.csv")
kc_house_preprocessed_df
```

```python
kcp_house_df = kc_house_preprocessed_df.drop(columns=["Unnamed: 0"])
kcp_house_df.head()
```

```python
kcp_house_df = kc_house_preprocessed_df.drop(columns=["Unnamed: 0"])
kcp_house_df.head()
```

```python
kcp_house_df.shape[0]
```

```python

from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_absolute_error
from sklearn.metrics import mean_squared_error
import numpy as np
from sklearn.metrics import mean_squared_error
from sklearn.metrics import r2_score
#np.sqrt(mean_squared_error(y_pred, y_true))
#mse = mean_squared_error(y_pred, y_true)
#mae = mean_absolute_error(y_pred, y_true)
#r2_score(y_pred, y_true)
```

```python
X = kcp_house_df.drop("price", axis=1)
y = kcp_house_df["price"]

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size = 0.3, random_state = 42)
```

```python
lin_reg = LinearRegression()
lin_reg.fit(X_train, y_train)

y_pred = lin_reg.predict(X_test)
```

```python
lin_reg.score(X_test, y_test)
```

```python
# Predict prices for the training rows that the model learned from.
# Compare these predictions with the test-set predictions to check generalization.
y_pred_train = lin_reg.predict(X_train)
```

```python
# MAE is the average absolute prediction error, measured in dollars.
# Compare training and test MAE using predictions from their matching datasets.
mae_train = mean_absolute_error(y_train, y_pred_train)
mae_test = mean_absolute_error(y_test, y_pred)
print(f"Training MAE: ${mae_train:,.2f}")
print(f"Test MAE: ${mae_test:,.2f}")
```

```python
# MSE averages squared errors, so large prediction errors count more heavily.
mse_train = mean_squared_error(y_train, y_pred_train)
mse_test = mean_squared_error(y_test, y_pred)
print(f"Training MSE: {mse_train:,.2f}")
print(f"Test MSE: {mse_test:,.2f}")
```

```python
# RMSE is the square root of MSE and is expressed in the target's units (dollars).
rmse_train = np.sqrt(mse_train)
rmse_test = np.sqrt(mse_test)
print(f"Training RMSE: ${rmse_train:,.2f}")
print(f"Test RMSE: ${rmse_test:,.2f}")
```

```python
# R2 measures how much better predictions are than using the split's mean price.
# A higher value is better; compare train and test to assess generalization.
r2_train = r2_score(y_train, y_pred_train)
r2_test = r2_score(y_test, y_pred)
print(f"Training R2: {r2_train:.4f}")
print(f"Test R2: {r2_test:.4f}")
```

```python
# Adjusted R2 penalizes R2 for the number of predictors in the model.
def adjusted_r2_score(r2, n, k):
    return 1 - ((1 - r2) * (n - 1) / (n - k - 1))

# Use each split's sample count and the number of input features.
number_of_features = X_train.shape[1]
adjusted_r2_train = adjusted_r2_score(
    r2_train, len(y_train), number_of_features
)
adjusted_r2_test = adjusted_r2_score(
    r2_test, len(y_test), number_of_features
)
print(f"Training adjusted R2: {adjusted_r2_train:.4f}")
print(f"Test adjusted R2: {adjusted_r2_test:.4f}")
```

## Why Predict on the Training Data?

`y_pred_train = lin_reg.predict(X_train)` asks the trained model to estimate prices for the same houses it used during fitting. These are predictions, not the actual prices: the actual training prices are in `y_train`. The model learns one linear equation that minimizes errors across the training examples, so it does not necessarily memorize or exactly reproduce every house price. `price` is excluded from `X_train` because it is the value the model is supposed to predict.

We calculate training metrics to measure how well the model fits the examples it learned from. We calculate test metrics by comparing `y_pred` with `y_test`; the test houses were held out during fitting, so these metrics give a more realistic estimate of performance on unseen houses. A much better training score than test score can signal overfitting. Similar scores are reassuring, but do not by themselves prove that the model is accurate enough.

### Results on This Split

- Training and test **R2** are both about **0.70**. The model explains roughly 70% of the price variation in each split; this is not 70% prediction accuracy.
- **MAE** is about **$124,692** on training data and **$127,487** on test data. Predictions miss the true price by about these amounts on average, using absolute error.
- **RMSE** is about **$198,134** on training data and **$208,297** on test data. RMSE is in dollars and penalizes large errors more than MAE, so its being higher suggests some predictions have larger misses.

The train and test results are fairly close, so there is no obvious large train/test performance gap here. However, the dollar errors are substantial; inspect residuals and compare against a simple baseline before deciding whether this model is useful.
